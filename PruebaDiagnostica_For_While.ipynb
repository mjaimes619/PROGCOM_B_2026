```python
## UNO FOR
numero = int(input("Ingresa un número para ver su tabla: "))

for i in range(1, 11):
    print(numero, "x", i, "=", numero * i)
```

```python
## UNO WHILE

numero = int(input("Ingresa un número para ver su tabla: "))

i = 1
while i <= 10:
    print(numero, "x", i, "=", numero * i)
    i += 1
```

```python
##DOS FOR
n = int(input("Hasta qué número quieres sumar: "))
suma = 0

for i in range(1, n + 1):
    suma += i

print("La suma es:", suma)
```

```python
##DOS WHILE
n = int(input("Hasta qué número quieres sumar: "))
suma = 0
i = 1

while i <= n:
    suma += i
    i += 1

print("La suma es:", suma)
```

```python
##TRES FOR
n = int(input("Ingresa un número para calcular su factorial: "))

if n < 0:
    print("No existen factoriales negativos")
else:
    resultado = 1
    for i in range(1, n + 1):
        resultado *= i
    print(n, "! =", resultado)
```

```python
##TRES WHILE
n = int(input("Ingresa un número para calcular su factorial: "))

if n < 0:
    print("No existen factoriales negativos")
else:
    resultado = 1
    i = 1
    while i <= n:
        resultado *= i
        i += 1
    print(n, "! =", resultado)
```

```python
##CUATRO FOR
n = int(input("Cuántos números de Fibonacci quieres ver: "))
a = 0
b = 1

for i in range(n):
    print(a, end=" ")
    temp = a + b
    a = b
    b = temp
print()
```

```python
##CUATRO WHILE
n = int(input("Cuántos números de Fibonacci quieres ver: "))
a = 0
b = 1
i = 0

while i < n:
    print(a, end=" ")
    temp = a + b
    a = b
    b = temp
    i += 1
print()
```

```python
##CINCO FOR
print("Ingresa 10 colores:")
colores = []
for i in range(10):
    color = input("Color " + str(i + 1) + ": ")
    colores.append(color)

print("\nAhora intenta adivinar uno (tienes 10 intentos):")
adivinado = False

for i in range(10):
    intento = input("Intento " + str(i + 1) + ": ")
    if intento in colores:
        print("¡Correcto!")
        adivinado = True
        break

if not adivinado:
    print("Se acabaron los intentos. Los colores eran:", colores)
```

```python
## CINCO WHILE
print("Ingresa 10 colores:")
colores = []
i = 0
while i < 10:
    color = input("Color " + str(i + 1) + ": ")
    colores.append(color)
    i += 1

print("\nAhora intenta adivinar uno (tienes 10 intentos):")
adivinado = False
i = 0

while i < 10:
    intento = input("Intento " + str(i + 1) + ": ")
    if intento in colores:
        print("¡Correcto!")
        adivinado = True
        break
    i += 1

if not adivinado:
    print("Se acabaron los intentos. Los colores eran:", colores)
```

```python
## SEIS FOR
for i in range(1, 101):
    if i % 2 == 0:
        print(i, end=" ")
print()
```

```python
## SEIS WHILE
i = 1
while i <= 100:
    if i % 2 == 0:
        print(i, end=" ")
    i += 1
print()
```

```python
## SIETE FOR
numero = input("Ingresa un número: ")
suma = 0

for digito in numero:
    suma += int(digito)

print("La suma de los dígitos es:", suma)
```

```python
## SIETE WHILE
numero = input("Ingresa un número: ")
suma = 0
i = 0

while i < len(numero):
    suma += int(numero[i])
    i += 1

print("La suma de los dígitos es:", suma)
```

```python
## OCHO FOR
n = int(input("Ingresa un número: "))
es_primo = True

if n < 2:
    es_primo = False
else:
    for i in range(2, n):
        if n % i == 0:
            es_primo = False
            break

if es_primo:
    print(n, "es primo")
else:
    print(n, "no es primo")
```

```python
## OCHO WHILE
n = int(input("Ingresa un número: "))
es_primo = True

if n < 2:
    es_primo = False
else:
    i = 2
    while i < n:
        if n % i == 0:
            es_primo = False
            break
        i += 1

if es_primo:
    print(n, "es primo")
else:
    print(n, "no es primo")
```

```python
## NUEVE FOR
filas = int(input("Cuántas filas quieres: "))

for i in range(1, filas + 1):
    for j in range(i):
        print("*", end="")
    print()
```

```python
## NUEVE WHILE
filas = int(input("Cuántas filas quieres: "))
i = 1

while i <= filas:
    j = 0
    while j < i:
        print("*", end="")
        j += 1
    print()
    i += 1
```

```python
## DIEZ FOR
base = int(input("Ingresa la base: "))
exponente = int(input("Ingresa el exponente: "))
resultado = 1

for i in range(exponente):
    resultado *= base

print(base, "^", exponente, "=", resultado)
```

```python
## DIEZ WHILE
base = int(input("Ingresa la base: "))
exponente = int(input("Ingresa el exponente: "))
resultado = 1
i = 0

while i < exponente:
    resultado *= base
    i += 1

print(base, "^", exponente, "=", resultado)
```
