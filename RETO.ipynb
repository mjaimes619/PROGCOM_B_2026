```python
menu = ["Combo 1","Combo 2","Combo 3","Combo 4","Papas","Jugos naturales","Cerveza","Agua","Gaseosas"]
precio = [20000, 25000, 18000, 24500, 3000, 7000, 3500, 4000, 2500]

usuario = input("Hay clientes? (s/n): ")
if usuario == "s":
    nuevo_cliente = "s"
    while nuevo_cliente == "s":
        pagar = 0  # ← se reinicia por cada cliente nuevo
        pedido = input("Desea ordenar? (s/n): ")
        while pedido == "s":
            ans = input("Que desea pedir: ")
            if ans in menu:
                cantidad = int(input(f"Cuantas unidades de '{ans}' desea?: "))
                subtotal = precio[menu.index(ans)] * cantidad
                pagar += subtotal
                print(f"  {cantidad}x {ans} = ${subtotal:,}")
            else:
                print("Producto no existente")

            pedido = input("Desea ordenar algo mas? (s/n): ")

        print(f"\nTotal a pagar: ${pagar:,}")
        print("-" * 30)
        nuevo_cliente = input("Hay un nuevo cliente? (s/n): ")

    print("No hay mas clientes. Hasta luego!")
else:
    print("No hay clientes")
```
