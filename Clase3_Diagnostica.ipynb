```python
a=float(input(" Ingrese el a de la ecuacion: "))
b=float(input(" Ingrese el b de la ecuacion: "))
c=float(input(" Ingrese el c de la ecuacion: "))
x1=(-b+(((b**2)-(4*a*c))**0.5))/2*a
x2=(-b-(((b**2)-(4*a*c))**0.5))/2*a

print(f"x1 es {x1} x2 es {x2}")
```
