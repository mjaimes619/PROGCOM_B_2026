#ESTRUCTURA DE REPETICION
1. for. cuando se sabe cuantas veces se debe repetir una acción.
2. while. No se sabe cuantas veces se repite la acción.

##1. FOR
Sintaxis general en python
```
for(variable)
  se repite una acción
```
algo

* in range(n): quiero repetir n veces
* in range(C1,CF,p): Quiero repetir desde CI hasta CF-1, con paso p
* in (lista,diccionario): Cuando quiero iterar cada elemento dentro de la lista/diccionario
* for each(lista, diccionario): Cuando quiero iterar cada elemento dentro de la lista/diccionario


````python
#Quiero repetir n veces
for i in range(9):
  print(f"Hay {i+1} personas en el salon")
````

````python
for i in range(1,10):
  print(f"Hay {i} personas en el salon")
````

````python
dias=int(input("Ey!! Cuántos dias falta para tu cumpleaños!? "))
for cumple in range(dias,0,-1):
  print(f"Faltan {cumple} dias para tu cumpleaños")
print("HOY ES TU CUMPLE ")
print("FELIZ CUMPLEAÑOS!! ")
````

#WHILE
Sintaxis

```
while(C1):
 lo que se va a repetir
 modificador(C1)
```

````python
c1=0
c2=0
c3=0
c4=0
nulo=0
votos="s"
while votos=="s":
  voto=int(input("Ingrese el número de su candidato (1-3).\n0 para voto en blanco "))
  if voto==1: c1=c1+1
  elif voto==2: c2+=1
  elif voto==3: c3+=1
  elif voto==0: c4+=1
  else: nulo+=1
  votos=input("Hay mas votos? s para si, n para no ")
  votos=votos[0].lower()
  print("")
print(f"Total de votos:\nCandidato 1={c1}\nCandidato 2={c2}\nCandidato 3={c3}\nVoto en blanco={c4}\nNulo={nulo}")
````

````python
##16. WHILE
numero = int(input("Ingresa un número para ver su tabla de multiplicar: "))
contador = 1
while contador <= 10:
    resultado = numero * contador
    print(f"{numero} x {contador} = {resultado}")
    contador += 1
````

````python
##16. FOR
numero=int(input("Ingresa el numero del que necesites la tabla: "))

for i in range(1,11):
  resultado=numero*i
  print(f"{numero} X {i}= {resultado}")
````

````python
numero = int(input("Ingresa la cantidad de numeros a sumar: "))
suma=0
contador = 1
while contador <= numero:
  suma += contador
  print(f"Sumando {contador}... (Suma parcial: {suma})")
  contador += 1
````

````python
numero = int(input("Ingresa la cantidad de numeros a sumar: "))
suma=0
contador = 1
for contador in range (1,numero+1):
  suma += contador
  print(f"Sumando {contador}... (Suma parcial: {suma})")
  contador += 1
````
