```python
def shift_derecha():
    print("\nShift derecha:")
    print("* Peso hacia la derecha")
    print("* Cabeza gira a la derecha")
    print("* Brazo izquierdo en forma de W relajada")


def brazo_arriba(lado):
    print(f"* Brazo {lado} extendido en diagonal")
    print("* Brazo opuesto abajo")


def alternar_brazos(n):
    print(f"\nAlternar brazos ({n} veces):")
    for i in range(n):
        print(f"  Repetición {i+1}:")
        if i % 2 == 0:
            brazo_arriba("izquierdo")
        else:
            brazo_arriba("derecho")


def saltos(n):
    print(f"\nSaltos mirando arriba ({n} veces):")
    for i in range(n):
        print(f"* Salto {i+1} mirando suavemente hacia arriba")


def caderas_lateral(n):
    print(f"\nCaderas izquierda-derecha ({n} veces):")
    print("* Ambos brazos arriba y extendidos")
    for i in range(n):
        lado = "izquierda" if i < n / 2 else "derecha"
        print(f"* Movimiento hacia {lado}")


def manos_cabeza(n):
    print(f"\nManos a nivel cabeza ({n} veces):")
    for i in range(n):
        lado = "izquierda" if i % 2 == 0 else "derecha"
        print(f"* Ambas manos hacia {lado}")


def paso_derecha_rodillas(n):
    print(f"\nPaso derecha con rodillas ({n} veces):")
    print("* Paso a la derecha")
    print("* Rodillas hacia adentro")
    print("* Brazos rígidos en diagonal")
    for i in range(n):
        lado = "izquierda" if i < n / 2 else "derecha"
        print(f"* Caderas hacia {lado}")


def brazos_frente_lateral(n):
    print(f"\nBrazos al frente y lateral ({n} veces):")
    print("* Pies juntos")
    for i in range(n):
        print(f"  Repetición {i+1}:")
        print("* Brazos al frente")
        print("* Mover a izquierda")
        print("* Mover a derecha")


def variante_pie(n):
    print(f"\nVariante con elevación de pie ({n} veces):")
    for i in range(n):
        print(f"  Repetición {i+1}:")
        print("* Elevar pie contrario al lado de brazos")
        print("* Brazos se mueven lateralmente detrás del torso")


def brazos_cruzados():
    print("\nBrazos cruzados:")
    print("* Manos en hombros")
    print("* Mano izquierda a mejilla izquierda")
    print("* Mano derecha a mejilla derecha")


def pose_final():
    print("\nPose final:")
    print("* Una mano en la cadera")
    print("* Otra extendida arriba")
    print("* Pierna levantada hacia atrás")
```

```python
shift_derecha()
alternar_brazos(4)
saltos(4)
caderas_lateral(4)
manos_cabeza(4)
paso_derecha_rodillas(4)
brazos_frente_lateral(2)
variante_pie(2)
brazos_cruzados()
pose_final()
```
