```python
class Buffet:
  #Constructor
  def __init__(self, carbohidrato, proteina, fruta, vegetal, bebida):
    self.carbohidrato=carbohidrato
    self.proteina=proteina
    self.fruta=fruta
    self.vegetal=vegetal
    self.bebida=bebida

  def elegir_comida(self, plato, cubiertos, servilleta, cant):
    desayuno={"proteína":["huevo", "salchicha", "pollo", "carne"],
              "carbohidrato":["arepa","pancake", "pan"],
              "fruta":["banano","fresa", "mango", "uva"],
              "vegetal":["cebolla","aguacate","zanahoria"],
              "bebida":["gaseosas", "café", "chocolate"]}
    pedido=[self.carbohidrato, self.proteina, self.fruta, self.vegetal, self.bebida]
    comida=["carbohidrato", "proteína", "fruta", "vegetal", "bebida"]

    for i in range(len(pedido)):
      if pedido[i] in desayuno[comida[i]]:
        print(f"Ingreso {pedido[i]} en el plato")
      else:
        print("Elija otro ingrediente")
```

```python
    if self.proteina in desayuno["proteína"]:
      print("Meto la", self.proteina, "en el plato")
    else:
      print("La proteína no está disponible")

    if self.carbohidrato in desayuno["carbohidrato"]:
        print(f"Meto el {self.carbohidrato} en el plato")
    else:
        print("El carbohidrato no está disponible")

    if self.fruta in desayuno["fruta"]:
        print(f"Meto la {self.fruta} en el plato")
    else:
        print("La fruta no está disponible")

    if self.vegetal in desayuno["vegetal"]:
        print(f"Meto el {self.vegetal} en el plato")
    else:
        print("El vegetal no está disponible")

    if self.bebida in desayuno["bebida"]:
        print(f"Agrego el {self.bebida} al pedido")
    else:
        print("La bebida no está disponible")
```
